```sql
USE CATALOG main;
USE nyc_taxi;

CREATE OR REPLACE VIEW vw_gold_daily AS

SELECT
  pickup_date,
  year,
  month,
  vendor_id,
  payment_type,
  n_trips,
  avg_fare,
  avg_tip_rate,
  dayofweek(pickup_date) AS dow,
  date_format(pickup_date, 'yyyy-MM-dd') AS dstr
FROM yellow_trip_2016_02_gold_daily;
```
